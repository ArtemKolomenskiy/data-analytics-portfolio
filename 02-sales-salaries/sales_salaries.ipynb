## **2. Построение витрины данных**

```python
paid = (payments
        .merge(managers, on='manager_id', validate='many_to_one')
        .merge(heads,    on='head_id',    validate='many_to_one')
        .merge(products, on='product_id', validate='many_to_one'))
paid = paid.loc[paid.invoice_status.isin(['Оплачен', 'Частично оплачен'])]
paid = paid.loc[paid.payment_date >= PAYMENT_CUTOFF]
paid['accrual_year']    = paid.payment_date.dt.year
paid['accrual_month']   = paid.payment_date.dt.month
paid['accrual_quarter'] = paid.payment_date.dt.quarter
assert paid.payment_id.is_unique

ACC_COLS = ['employee_role', 'employee_id', 'employee_name',
            'head_id', 'head_name', 'department_name',
            'accrual_year', 'accrual_month', 'accrual_quarter',
            'accrual_type', 'product_name',
            'deal_id', 'invoice_id', 'payment_id', 'payment_date',
            'calculation_base', 'rate_percent', 'accrual_amount']

sale_bonus = pd.DataFrame({
    'employee_role':    'manager',
    'employee_id':      paid.manager_id,
    'employee_name':    paid.manager_name,
    'head_id':          paid.head_id,
    'head_name':        paid.head_name,
    'department_name':  paid.department_name,
    'accrual_year':     paid.accrual_year,
    'accrual_month':    paid.accrual_month,
    'accrual_quarter':  paid.accrual_quarter,
    'accrual_type':     'Премия',
    'product_name':     paid.product_name,
    'deal_id':          paid.deal_id,
    'invoice_id':       paid.invoice_id,
    'payment_id':       paid.payment_id,
    'payment_date':     paid.payment_date,
    'calculation_base': paid.payment_amount,
    'rate_percent':     10.0,
    'accrual_amount':   paid.payment_amount * 0.10,
})

team_q = (paid.groupby(['head_id', 'head_name', 'accrual_year', 'accrual_quarter'],
                       as_index=False)
                .agg(team_sales=('payment_amount', 'sum')))
quarter_bonus = pd.DataFrame({
    'employee_role':    'head',
    'employee_id':      team_q.head_id,
    'employee_name':    team_q.head_name,
    'head_id':          team_q.head_id,
    'head_name':        team_q.head_name,
    'department_name':  None,
    'accrual_year':     team_q.accrual_year,
    'accrual_month':    team_q.accrual_quarter * 3,
    'accrual_quarter':  team_q.accrual_quarter,
    'accrual_type':     'Квартальная премия',
    'product_name':     'Квартальная премия',
    'deal_id':          None,
    'invoice_id':       None,
    'payment_id':       None,
    'payment_date':     pd.NaT,
    'calculation_base': team_q.team_sales,
    'rate_percent':     1.0,
    'accrual_amount':   team_q.team_sales * 0.01,
})

months_df = pd.DataFrame({'month_start': REPORT_STARTS, '_k': 1})
head_salary = (heads.assign(_k=1).merge(months_df, on='_k').drop(columns='_k')
                    .rename(columns={'month_start': 'salary_month'}))
head_salary_df = pd.DataFrame({
    'employee_role':    'head',
    'employee_id':      head_salary.head_id,
    'employee_name':    head_salary.head_name,
    'head_id':          head_salary.head_id,
    'head_name':        head_salary.head_name,
    'department_name':  None,
    'accrual_year':     head_salary.salary_month.dt.year,
    'accrual_month':    head_salary.salary_month.dt.month,
    'accrual_quarter':  head_salary.salary_month.dt.quarter,
    'accrual_type':     'Оклад',
    'product_name':     'Оклад',
    'deal_id':          None,
    'invoice_id':       None,
    'payment_id':       None,
    'payment_date':     pd.NaT,
    'calculation_base': head_salary.salary_amount,
    'rate_percent':     100.0,
    'accrual_amount':   head_salary.salary_amount,
})

mgr_salary = (managers.assign(_k=1).merge(months_df, on='_k')
                       .merge(heads, on='head_id', suffixes=('', '_h'))
                       .drop(columns='_k')
                       .rename(columns={'month_start': 'salary_month'}))
mgr_salary_df = pd.DataFrame({
    'employee_role':    'manager',
    'employee_id':      mgr_salary.manager_id,
    'employee_name':    mgr_salary.manager_name,
    'head_id':          mgr_salary.head_id,
    'head_name':        mgr_salary.head_name,
    'department_name':  mgr_salary.department_name,
    'accrual_year':     mgr_salary.salary_month.dt.year,
    'accrual_month':    mgr_salary.salary_month.dt.month,
    'accrual_quarter':  mgr_salary.salary_month.dt.quarter,
    'accrual_type':     'Оклад',
    'product_name':     'Оклад',
    'deal_id':          None,
    'invoice_id':       None,
    'payment_id':       None,
    'payment_date':     pd.NaT,
    'calculation_base': mgr_salary.salary_amount,
    'rate_percent':     100.0,
    'accrual_amount':   mgr_salary.salary_amount,
})

salary = pd.concat([head_salary_df, mgr_salary_df], ignore_index=True)

accruals = (pd.concat([sale_bonus, quarter_bonus, salary], ignore_index=True)[ACC_COLS]
              .sort_values(['head_name', 'employee_role', 'employee_name',
                            'accrual_year', 'accrual_month', 'accrual_type'])
              .reset_index(drop=True))

assert set(accruals.accrual_type) == {'Премия', 'Квартальная премия', 'Оклад'}
assert accruals.loc[accruals.accrual_type.eq('Оклад'), 'accrual_amount'].ge(0).all()
assert abs(accruals.loc[accruals.accrual_type.eq('Премия'), 'accrual_amount'].sum()
           - paid.payment_amount.sum() * 0.10) < .01
assert abs(accruals.loc[accruals.accrual_type.eq('Квартальная премия'), 'accrual_amount'].sum()
           - paid.payment_amount.sum() * 0.01) < .01

display(accruals.head(12))
```

## **3. Построение агрегатов и сводной таблицы для дашборда**

```python
report = accruals.copy()
report['Руководитель']    = report.head_name
report['Менеджер']        = report.employee_name
report['Сумма продаж']    = report.calculation_base.where(report.accrual_type.eq('Премия'), 0.)
report['Премия']          = report.accrual_amount.where(
                                report.accrual_type.isin(['Премия', 'Квартальная премия']), 0.)
report['Оклад']           = report.accrual_amount.where(report.accrual_type.eq('Оклад'), 0.)
report['Всего начислено'] = report['Премия'] + report['Оклад']

metrics = ['Сумма продаж', 'Премия', 'Оклад', 'Всего начислено']
month_ru = {1:'Январь', 2:'Февраль', 3:'Март', 4:'Апрель', 5:'Май', 6:'Июнь',
            7:'Июль', 8:'Август', 9:'Сентябрь', 10:'Октябрь', 11:'Ноябрь', 12:'Декабрь'}

pivot = report.pivot_table(
    index=['Руководитель', 'Менеджер'],
    columns=['accrual_year', 'accrual_month'],
    values=metrics, aggfunc='sum', fill_value=0.,
)
pivot = pivot.reorder_levels([1, 2, 0], axis=1)
ordered = [(y, m, metric)
           for y, m in sorted({(y, m) for y, m, _ in pivot.columns})
           for metric in metrics]
pivot = pivot.reindex(columns=pd.MultiIndex.from_tuples(ordered), fill_value=0.)
pivot.columns = pd.MultiIndex.from_tuples(
    [(str(y), month_ru[m], metric) for y, m, metric in pivot.columns],
    names=['Год', 'Месяц', 'Показатель'],
)

totals = report.groupby(['Руководитель', 'Менеджер'])[metrics].sum()
for metric in metrics:
    pivot[('Общий итог', '', metric)] = totals[metric]
pivot.loc[('Общий итог', ''), :] = pivot.sum(axis=0)
pivot = pivot.fillna(0.)

assert abs(pivot.loc[('Общий итог', ''), ('Общий итог', '', 'Всего начислено')]
           - accruals.accrual_amount.sum()) < .01

def money(v): return '—' if abs(v) < .005 else f'{v:,.0f}'.replace(',', ' ')

display(pivot)
```

## **4. Формирование дашборда**

```python
from IPython.display import HTML, display
from html import escape
import uuid


def ru_int(v):
    return f'{int(v):,}'.replace(',', ' ')


def money(v):
    return '—' if abs(v) < 0.005 else f'{v:,.0f}'.replace(',', ' ')


month_ru_inv = {v: k for k, v in month_ru.items()}
uid_prefix = f"det-{uuid.uuid4().hex[:6]}"

emp_lookup = (report[['employee_name', 'employee_id', 'accrual_year', 'accrual_month']]
              .drop_duplicates())
emp_map = {
    (r.employee_name, r.accrual_year, r.accrual_month): r.employee_id
    for r in emp_lookup.itertuples()
}


def breakdown_rows_html(employee_id, year, month, metric):
    sub = report[
        (report['employee_id'] == employee_id)
        & (report['accrual_year'] == year)
        & (report['accrual_month'] == month)
    ]
    if metric == 'Сумма продаж':
        sub = sub[sub['accrual_type'] == 'Премия']
    elif metric == 'Премия':
        sub = sub[sub['accrual_type'].isin(['Премия', 'Квартальная премия'])]
    else:
        return '<p>Нет данных.</p>'

    if sub.empty:
        return '<p>Нет данных.</p>'

    rows = []
    for _, r in sub.iterrows():
        pay_date = r.payment_date.strftime('%d.%m.%Y') if pd.notna(r.payment_date) else '—'
        rows.append(
            '<tr>'
            f'<td>{escape(str(r.accrual_type))}</td>'
            f'<td>{escape(str(r.product_name or "—"))}</td>'
            f'<td>{escape(str(r.deal_id or "—"))}</td>'
            f'<td>{escape(str(r.invoice_id or "—"))}</td>'
            f'<td>{pay_date}</td>'
            f'<td class="num">{money(r.calculation_base)}</td>'
            f'<td class="num">{r.rate_percent:.0f}%</td>'
            f'<td class="num">{money(r.accrual_amount)}</td>'
            '</tr>'
        )
    return (
        '<table class="drill-table"><thead><tr>'
        '<th>Тип начисления</th><th>Продукт</th><th>Сделка</th><th>Счёт</th>'
        '<th>Дата платежа</th><th>Сумма платежа</th><th>Процент</th><th>Сумма премии</th>'
        '</tr></thead><tbody>' + ''.join(rows) + '</tbody></table>'
    )


def build_pivot_html_with_drill(pivot_df, money_fmt):
    pivot_disp = pivot_df.copy().astype(object)
    modals = []
    uid_counter = [0]

    for col in pivot_df.columns:
        year_str, month_name, metric = col
        for idx in pivot_df.index:
            leader, manager = idx
            is_total_row = (leader == 'Общий итог')

            value = pivot_df.loc[idx, col]
            value_text = money_fmt(value)

            can_drill = (
                not is_total_row
                and metric in ('Сумма продаж', 'Премия')
                and value_text != '—'
                and str(year_str).isdigit()
            )

            if can_drill:
                year = int(year_str)
                month = month_ru_inv.get(month_name)
                emp_id = emp_map.get((manager, year, month))
                if emp_id is not None:
                    uid = f'{uid_prefix}-{uid_counter[0]}'
                    uid_counter[0] += 1

                    modals.append(
                        f'<div class="modal-container">'
                        f'<input type="checkbox" id="{uid}" class="drill-toggle">'
                        f'<div class="modal">'
                        f'<label for="{uid}" class="modal-backdrop"></label>'
                        f'<div class="modal-content">'
                        f'<label for="{uid}" class="modal-close">×</label>'
                        f'<h3>{escape(str(manager))} · {escape(month_name)} {year} · '
                        f'{escape(metric)}</h3>'
                        f'{breakdown_rows_html(emp_id, year, month, metric)}'
                        f'</div></div></div>'
                    )
                    pivot_disp.loc[idx, col] = (
                        f'<label for="{uid}" class="drill-link">{value_text}</label>'
                    )
                    continue

            pivot_disp.loc[idx, col] = value_text

    html = pivot_disp.to_html(
        classes='salary-pivot', border=0, escape=False, na_rep='—'
    )
    return html, modals


def build_salary_dashboard():
    total_bonus = int(report['Премия'].sum())
    total_salary = int(report['Оклад'].sum())
    total_accruals = int(report['Всего начислено'].sum())

    css = """<style>
    :root {
        --ink:      #0f1d33;
        --muted:    #5c6b82;
        --line:     #d6dee8;
        --bg:       #f4f6fa;
        --navy:     #1e3a5f;
        --blue-3:   #8fa9c7;
        --alert-bg: #e3ebf4;
    }

    .dashboard {
        font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI',
                     Roboto, 'Helvetica Neue', Arial, sans-serif;
        color: var(--ink);
        background: var(--bg);
        padding: 24px;
        border-radius: 4px;
        max-width: 1500px;
        border: 1px solid var(--line);
    }

    .dashboard h1 {
        font-size: 22px;
        font-weight: 600;
        letter-spacing: -0.01em;
        color: var(--navy);
        margin: 0 0 24px;
    }

    .kpis {
        display: flex;
        gap: 0;
        flex-wrap: wrap;
        margin-bottom: 24px;
        border: 1px solid var(--line);
        border-radius: 4px;
        overflow: hidden;
        background: #fff;
    }

    .kpi {
        background: #fff;
        padding: 16px 20px;
        min-width: 200px;
        flex: 1;
        border-right: 1px solid var(--line);
        position: relative;
    }

    .kpi:last-child { border-right: none; }

    .kpi::before {
        content: "";
        position: absolute;
        left: 0;
        top: 0;
        bottom: 0;
        width: 3px;
        background: var(--blue-3);
    }

    .kpi b {
        display: block;
        font-size: 24px;
        font-weight: 600;
        color: var(--navy);
        font-variant-numeric: tabular-nums;
        margin: 4px 0 2px;
    }

    .kpi span {
        font-size: 12px;
        color: var(--muted);
    }

  #    .table-wrap {
  #   margin-top: 16px;
  #   overflow-x: auto;
  #   width: 100%;
  #   max-width: 100%;
  # }

  #   .salary-pivot {
  #       border-collapse: collapse;
  #       font-size: 11px;
  #       white-space: nowrap;
  #       font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI',
  #                    Roboto, sans-serif;
  #       width: max-content;
  #       margin: 16px;
  #   }

  #   .salary-pivot th,
  #   .salary-pivot td {
  #       border: 1px solid var(--line);
  #       padding: 6px 10px;
  #       text-align: right;
  #   }

  #   .salary-pivot th {
  #       background: var(--navy);
  #       color: #fff;
  #       font-weight: 600;
  #       font-size: 11px;
  #       letter-spacing: 0.03em;
  #       text-transform: uppercase;
  #       text-align: center;
  #       vertical-align: middle;
  #   }

  #   .salary-pivot tbody th {
  #       text-align: left;
  #       background: #f7f9fc;
  #       color: var(--navy);
  #       font-weight: 600;
  #       text-transform: none;
  #   }

  #   .salary-pivot tbody tr:nth-child(even) td { background: #fafbfd; }
  #   .salary-pivot tbody tr:hover td { background: #eef3f9; }

  #   .salary-pivot tbody tr:last-child th,
  #   .salary-pivot tbody tr:last-child td {
  #       font-weight: 700;
  #       background: var(--alert-bg);
  #       color: var(--navy);
  #       border-top: 2px solid var(--navy);
  #   }
      /* ── Таблица ── */
    .table-wrap {
        overflow-x: auto;
        max-width: 100%;
        border: 1px solid var(--line);
        border-radius: 4px;
        background: #fff;
        margin-top: 16px;
    }

    .salary-pivot {
        border-collapse: collapse;
        font: 13px -apple-system, BlinkMacSystemFont, 'Segoe UI',
              Roboto, sans-serif;
        white-space: nowrap;
        width: max-content;
    }

    .salary-pivot th,
    .salary-pivot td {
        border-bottom: 1px solid var(--line);
        border-right: 1px solid var(--line);
        padding: 9px 12px;
        text-align: right;
        vertical-align: middle;
    }

    .salary-pivot thead th {
        background: var(--navy);
        color: #fff;
        font-weight: 600;
        font-size: 12px;
        letter-spacing: 0.03em;
        text-transform: uppercase;
        # position: sticky;
        # top: 0;
        text-align: center;
    }

    .salary-pivot tbody th {
        background: #f7f9fc;
        color: var(--navy);
        font-weight: 600;
        text-align: left;
        text-transform: none;
    }

    .salary-pivot tbody tr:nth-child(even) td {
        background: #fafbfd;
    }
    .salary-pivot tbody tr:hover td {
        background: #eef3f9;
    }

    .salary-pivot tbody tr:last-child th,
    .salary-pivot tbody tr:last-child td {
        font-weight: 700;
        background: var(--alert-bg);
        color: var(--navy);
        border-top: 2px solid var(--navy);
    }

    .drill-toggle { display: none; }

    .drill-link {
        color: var(--navy);
        text-decoration: underline dotted;
        text-underline-offset: 2px;
        cursor: pointer;
    }
    .drill-link:hover { background: #eef3f9; }

    .modal {
        display: none;
        position: fixed;
        inset: 0;
        z-index: 9999;
        align-items: center;
        justify-content: center;
        padding: 24px;
    }

    .drill-toggle:checked ~ .modal { display: flex; }

    .modal-backdrop {
        position: absolute;
        inset: 0;
        background: rgba(15, 29, 51, 0.55);
        cursor: pointer;
    }

    .modal-content {
        position: relative;
        background: #fff;
        border-radius: 6px;
        padding: 24px 28px 20px;
        max-width: 1100px;
        max-height: 85vh;
        overflow-y: auto;
        box-shadow: 0 10px 40px rgba(0, 0, 0, 0.3);
        z-index: 1;
    }

    .modal-close {
        position: absolute;
        top: 8px;
        right: 14px;
        font-size: 26px;
        line-height: 1;
        text-decoration: none;
        color: var(--muted);
        cursor: pointer;
    }
    .modal-close:hover { color: var(--ink); }

    .modal h3 {
        margin: 0 0 12px;
        color: var(--navy);
        font-size: 16px;
        padding-right: 24px;
    }

    .drill-table {
        border-collapse: collapse;
        font-size: 12px;
        width: 100%;
    }
    .drill-table th,
    .drill-table td {
        border: 1px solid var(--line);
        padding: 6px 10px;
        text-align: left;
        white-space: nowrap;
    }
    .drill-table th {
        background: #f7f9fc;
        color: var(--navy);
        font-weight: 600;
    }
    .drill-table td.num {
        text-align: right;
        font-variant-numeric: tabular-nums;
    }
    </style>"""

    pivot_html, modals = build_pivot_html_with_drill(pivot, money)
    modals_html = ''.join(modals)

    kpi_html = f"""
    <div class="kpis">
        <div class="kpi">
            <span>Начислено по премиям</span>
            <b>{ru_int(total_bonus)} </b>
        </div>
        <div class="kpi">
            <span>Начислено по окладам</span>
            <b>{ru_int(total_salary)} </b>
        </div>
        <div class="kpi">
            <span>Начислено всего</span>
            <b style="color:#2b6777;">{ru_int(total_accruals)} </b>
        </div>
    </div>
    """

    return (
        css
        + '<div class="dashboard">'
        + '<h1>Заработные платы отдела продаж</h1>'
        + kpi_html
        + '<div class="table-wrap">' + pivot_html + '</div>'
        + modals_html
        + '</div>'
    )


display(HTML(build_salary_dashboard()))
```
